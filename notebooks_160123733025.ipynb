Q1. Load the COVID-19 vaccination dataset. Create a
Series of total vaccinations per country. Examine the
vaccination disparity between the top 10 and bottom
10 countries using index operations.

```python
import pandas as pd

df = pd.read_csv("country_vaccinations.csv")

total_vacc = df.groupby('country')['total_vaccinations'].max().sort_values(ascending=False)

top10 = total_vacc.head(10)
bottom10 = total_vacc.tail(10)

top10, bottom10
```

### Analysis

The total number of vaccinations per country reveals a significant disparity between nations. The top 10 countries contribute a disproportionately large share of global vaccinations, which can be attributed to factors such as larger population sizes, better healthcare infrastructure, and stronger vaccine procurement strategies.

In contrast, the bottom 10 countries show extremely low vaccination counts. This may be due to limited access to vaccines, logistical challenges, or smaller population sizes. The gap between the average vaccinations of the top and bottom groups highlights global inequality in vaccine distribution.

This disparity emphasizes the need for equitable vaccine allocation and international cooperation during global health crises.

Q2. Examine the relationship between
&#39;people_vaccinated&#39; and &#39;people_fully_vaccinated&#39;
using ufuncs such as np.log and np.sqrt. Assess
whether full vaccination rates are proportional to initial
vaccination rates.

```python
import numpy as np

df[['people_vaccinated','people_fully_vaccinated']] = \
df[['people_vaccinated','people_fully_vaccinated']].fillna(0)

df['log_vaccinated'] = np.log1p(df['people_vaccinated'])
df['sqrt_fully'] = np.sqrt(df['people_fully_vaccinated'])

correlation = df['people_vaccinated'].corr(df['people_fully_vaccinated'])
correlation
```

### Analysis

The relationship between `people_vaccinated` and `people_fully_vaccinated` is examined using NumPy universal functions such as logarithmic and square root transformations. These transformations help reduce skewness in the data, as vaccination counts vary widely across countries.

The computed correlation coefficient is very high (close to 1), indicating a strong positive relationship between the two variables. This suggests that countries that vaccinate more people initially also tend to have higher numbers of fully vaccinated individuals.

However, the relationship is not perfectly proportional. This is because full vaccination typically requires multiple doses and involves time delays, policy differences, and supply constraints. Therefore, while strongly related, the progression from initial vaccination to full vaccination is influenced by multiple real-world factors.

Q3. Apply hierarchical indexing on country and date.
Extract weekly vaccination progress for three selected
countries using xs(). Examine trends at multiple index
levels.

```python
df['date'] = pd.to_datetime(df['date'])

df_multi = df.set_index(['country','date']).sort_index()

weekly = df_multi.groupby(level=0).resample('W', level=1).sum()

weekly.xs('India', level='country').head()
```

### Analysis

Hierarchical (multi-level) indexing using country and date enables more structured and flexible data analysis. By organizing the dataset in this way, it becomes easier to perform time-series operations within each country.

The use of weekly resampling smooths daily fluctuations and provides a clearer view of vaccination trends over time. Extracting specific countries using the `xs()` function allows focused analysis without modifying the overall dataset structure.

From the weekly trends, it is possible to observe differences in vaccination rollout speeds, periods of rapid growth, and potential slowdowns. This multi-level indexing approach is especially useful for comparing trends across different countries while preserving time-based granularity.

Q5. Identify columns with more than 30% missing
values. Examine whether to drop them or impute.
Apply forward-fill (ffill) for time-series columns and
justify this choice with evidence.

```python
missing = df.isnull().mean()*100

high_missing = missing[missing > 30]

df_sorted = df.sort_values(['country','date'])
df_filled = df_sorted.groupby('country').ffill()

high_missing
```

### Analysis

Columns with more than 30% missing values indicate potential data quality issues and require careful handling. Simply dropping these columns may lead to loss of valuable information, especially if they contain important vaccination metrics.

For time-series data such as vaccination counts, forward-fill (`ffill`) is a suitable imputation method. Since vaccination data is cumulative in nature, missing values often represent delays in reporting rather than actual decreases.

By using forward-fill, the last known value is propagated forward, preserving the continuity of the dataset. This approach ensures that trends remain realistic and avoids introducing artificial drops or inconsistencies that could distort analysis.

Q7. Design a daily vaccination rate calculator using
apply() and diff(). Combine results from multiple
countries into a unified hierarchical DataFrame and
highlight top-performing nations.

```python
df = df.sort_values(['country','date'])

df['daily_rate'] = df.groupby('country')['total_vaccinations'].diff()

top_daily = df.groupby('country')['daily_rate'].mean().sort_values(ascending=False)

top_daily.head(10)
```

### Analysis

The daily vaccination rate is calculated using the difference (`diff()`) of cumulative vaccination counts. This provides a measure of how many vaccinations are administered each day.

By grouping the data by country, we can compute and compare average daily vaccination rates across nations. Countries with higher average daily rates demonstrate stronger logistical capabilities, efficient distribution systems, and effective vaccination campaigns.

Combining these results into a hierarchical DataFrame allows for structured comparison across countries and dates. Identifying top-performing countries helps highlight best practices and successful vaccination strategies.

Q8. Examine the impact of using dropna() versus
fillna(method=&#39;ffill&#39;) on cumulative vaccination totals for
India and the USA. Compare summary statistics
before and after each approach.

```python
subset = df[df['country'].isin(['India','United States'])]

dropna_df = subset.dropna()
ffill_df = subset.groupby('country').ffill()

dropna_df.describe(), ffill_df.describe()
```

### Analysis

Handling missing data significantly impacts the results of the analysis. Using `dropna()` removes all rows with missing values, which can lead to substantial data loss, especially in time-series datasets.

In contrast, forward-fill (`ffill`) retains the dataset's structure by filling missing values with the last observed value. This is particularly appropriate for cumulative vaccination data, where values are expected to remain constant or increase over time.

Comparing summary statistics before and after applying these methods shows that `dropna()` may underestimate trends due to reduced data, while `ffill()` preserves continuity and provides a more accurate representation of vaccination progress.

Therefore, forward-fill is generally the preferred approach for this type of data.

Q10. Build a multi-country vaccination comparison
using pivot_table with hierarchical indexing on country
and vaccine type. Export a structured summary and
push the notebook to GitHub.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("country_vaccinations.csv")

# Step 1: Split vaccines into list
df['vaccines'] = df['vaccines'].str.split(',')

# Step 2: Explode so each vaccine gets its own row
df_exploded = df.explode('vaccines')

# Clean spaces
df_exploded['vaccines'] = df_exploded['vaccines'].str.strip()

# Step 3: Create pivot table
pivot = pd.pivot_table(
    df_exploded,
    values='total_vaccinations',
    index='country',
    columns='vaccines',
    aggfunc='max'
)

pivot.head()
```

### Analysis

The pivot table provides a multi-dimensional view of the dataset by organizing total vaccinations across countries and vaccine types. This structure enables easy comparison of how different vaccines are distributed globally.

Hierarchical indexing improves readability and allows users to analyze relationships between countries and the vaccines they use. It also helps identify which vaccines are most widely adopted and how vaccination efforts vary across regions.

Exporting the pivot table as a CSV file ensures that the results can be shared, reused, and integrated into reports or dashboards. This step is essential for reproducibility and collaboration, especially when publishing the analysis on platforms like GitHub.
